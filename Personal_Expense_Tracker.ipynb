# Personal Expense Tracker

```python
expenses = []

def add_expense(records):
    title = input("Expense description: ").strip()
    category = input("Category (food/travel/study/other): ").strip().lower() or "other"
    try:
        amount = float(input("Amount (₹): "))
        if amount <= 0:
            print("Amount must be greater than zero.")
            return
    except ValueError:
        print("Please enter a valid amount.")
        return

    records.append({"title": title, "category": category, "amount": amount})
    print("Expense added.")

def show_expenses(records):
    if not records:
        print("No expenses recorded yet.")
        return
    print("\\n--- EXPENSES ---")
    for i, item in enumerate(records, start=1):
        print(f"{i}. {item['title']} | {item['category']} | ₹{item['amount']:.2f}")
    print(f"Total spent: ₹{sum(item['amount'] for item in records):.2f}")

def category_summary(records):
    totals = {}
    for item in records:
        totals[item["category"]] = totals.get(item["category"], 0) + item["amount"]
    if not totals:
        print("No data to summarize.")
        return
    print("\\n--- CATEGORY SUMMARY ---")
    for category, total in sorted(totals.items()):
        print(f"{category.title():12} ₹{total:.2f}")

while True:
    print("\\n1. Add expense  2. View expenses  3. Category summary  4. Exit")
    choice = input("Choose an option: ").strip()
    if choice == "1":
        add_expense(expenses)
    elif choice == "2":
        show_expenses(expenses)
    elif choice == "3":
        category_summary(expenses)
    elif choice == "4":
        print("Tracker closed.")
        break
    else:
        print("Choose 1, 2, 3, or 4.")
```
