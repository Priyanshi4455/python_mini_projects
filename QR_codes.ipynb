```python
url = input("Enter URL to Make QR: ")
if url.startswith("http://") or url.startswith("https://"):
    qr = qrcode.make(url)
    qr.save("SecondQR.png")
    print("QR code created successfully!")
else:
    print("invalid url...!!! must start with http or https")
```

```python
url=input("Enter URL to Make QR: ")
if url.startswith("http://") or url.startswith("https://"):
    qr=qrcode.QRCode(version=1,
    error_correction=qrcode.constants. ERROR_CORRECT_H,
    box_size=10,# 10×10
    border=4
    )
    qr.add_data(url)
    qr.make()
    
    image=qr.make_image(fill_color="blue", back_color="red")
    
    image.save("ThirdQR.png")
    print("QR code created successfully!")
else:
    print("Invalid Url .. !!! Must start with http or https")
```

```python
#Create QR Code For UPI with sapcific Amount

upi_id=input("Enter UPI ID: ")
name=input("Enter Name: ")
amount=int(input("Enter Amount: "))

upi_url= f"upi://pay?pa={upi_id}&pn={name}&am={amount}"

#pa= Payee
#pn= Payee Name
#am= Amount
#cu= INR

qr=qrcode.make(upi_url)
qr.save("Priyanshi_UPI_2.png")

print("QR code created successfully!")
```

```python
#Text to Image    >>>(Image, ImageDraw, ImageFont)
from PIL import Image, ImageDraw, ImageFont
txt=input("Enter Your Text: ")
```

```python
#Create Image
image=Image.new("RGB", (800,300), "white")

#Create Drawing Object
draw=ImageDraw.Draw(image)

#Load Font
font=ImageFont.load_default()    #fontsize, fontStyle

#Draw text on image
draw.text( (50,100) , txt, fill="black",font=font)

#Save Image
image.save("txt_image.png")
print("Image Generated Successfully!")
```
